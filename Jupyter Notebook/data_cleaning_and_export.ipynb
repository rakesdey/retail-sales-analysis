```python
import io
import psycopg2


# Step 2: Connect
conn = psycopg2.connect(
    host="localhost",
    dbname="Project 01",
    user="postgres",
    password="123456",
    port="5432"
)
cur = conn.cursor()

query = """
SELECT 
    o.row_id, o.order_id, o.order_date, o.ship_date, o.ship_mode,
    o.city, o.state, o.postal_code, o.region,
    c.customer_id, c.customer_name, c.segment, c.country,
    p.product_id, p.category, p.sub_category, p.product_name,
    o.sales, o.quantity, o.discount, o.profit, o.shipping_days, o.profit_margin_pct
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON o.product_id = p.product_id;
"""

export_df = pd.read_sql(query, conn)
export_df.to_csv(r'D:\Project 1\Retail Analysis\superstore_export.csv', index=False)
conn.close()

print("Exported:", export_df.shape)
```
